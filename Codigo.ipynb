```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv('bicicletas_compartidas_bogota.csv')
```

**Punto 1**

```python
sub = df.dropna(subset=['distancia_km','duracion_min'])
x = sub['distancia_km'].to_numpy()
y = sub['duracion_min'].to_numpy()

cov = np.cov(x, y, ddof=1)[0, 1]

r = cov / (x.std(ddof=1) * y.std(ddof=1))
print(f"Pearson r: {r:.4f}")
```

**1.b** 

**Dirección:** positiva, a mayor distancia recorrida, tiende a haber mayor duración del viaje.

**Intensidad:** moderada-alta. Un r de 0.72 no es una correlación perfecta, pero sí es considerablemente fuerte para datos de comportamiento humano/urbano, donde hay mucha variabilidad.

**Significado práctico:** R²≈0.52. El 52% de la variación en duración se explica linealmente por la distancia recorrida. El 48% restante depende de otros factores no capturados aquí.

**1.c**

No es correcto afirmarlo únicamente con base en la correlación, porque esta muestra asociación, no causalidad. Aunque existe una relación física razonable entre distancia y duración, podrían influir otras variables como el clima o el tipo de bicicleta. Por eso, lo adecuado es decir que existe una asociación positiva entre ambas variables, no que la distancia causa directamente la duración.

**1.d**

No. Un coeficiente de correlación como r=0.72 resume una tendencia general calculada sobre miles de observaciones, pero no garantiza el comportamiento de cada viaje individual. Es un promedio estadístico, no una regla determinística.

**Justificación:** estos casos muestran que factores no capturados en la correlación pueden hacer que un viaje individual se aleje completamente del patrón general. La correlación describe cómo se comporta el conjunto de datos en promedio, pero viajes puntuales pueden romper esa tendencia sin que eso invalide el hallazgo global.

**Punto 2**

```python
q1_dist=df['distancia_km'].quantile(0.25)
q1_dur=df['duracion_min'].quantile(0.25)
print(q1_dist)
print(q1_dur)
```

```python
q3_dist=df['distancia_km'].quantile(0.75)
q3_dur=df['duracion_min'].quantile(0.75)
print(q3_dist)
print(q3_dur)
```

```python
iqr_dist=q3_dist-q1_dist
iqr_dur=q3_dur-q1_dur
print(iqr_dist)
print(iqr_dur)
```

```python
lim_inferior_dist = q1_dist - 1.5 * iqr_dist
lim_inferior_dur = q1_dur - 1.5 * iqr_dur
print(lim_inferior_dist)
print(lim_inferior_dur)
```

```python
lim_superior_dist = q3_dist + 1.5 * iqr_dist
lim_superior_dur = q3_dur + 1.5 * iqr_dur
print(lim_superior_dist)
print(lim_superior_dur)
```

```python
outliers_dist = sub[(sub['distancia_km'] < lim_inferior_dist) | (sub['distancia_km'] > lim_superior_dist)]
outliers_dur  = sub[(sub['duracion_min'] < lim_inferior_dur) | (sub['duracion_min'] > lim_superior_dur)]
print(outliers_dist)
print(outliers_dur)
```

```python
mask_outliers = ((sub['distancia_km'] < lim_inferior_dist) | (sub['distancia_km'] > lim_superior_dist) |
                  (sub['duracion_min'] < lim_inferior_dur) | (sub['duracion_min'] > lim_superior_dur))

outliers_totales = sub[mask_outliers]
print(f"Total de registros atípicos (en al menos una variable): {len(outliers_totales)} de {len(sub)} ({100*len(outliers_totales)/len(sub):.2f}%)")
```

```python
sub = sub.copy()

sub['velocidad_kmh'] = sub['distancia_km'] / (sub['duracion_min'] / 60)
sub['exceso_dist'] = (sub['distancia_km'] - lim_superior_dist).clip(lower=0) / iqr_dist
sub['exceso_dur']  = (sub['duracion_min'] - lim_superior_dur).clip(lower=0) / iqr_dur
sub['severidad'] = sub['exceso_dist'] + sub['exceso_dur']

top5 = sub.sort_values('severidad', ascending=False).head(5)
top5[['distancia_km', 'duracion_min', 'velocidad_kmh']]
```

Esto indica que el patrón observado está presente en el comportamiento general de los usuarios y no depende solamente de unos pocos registros atípicos.

Sin embargo, el cambio en el coeficiente muestra que la intensidad de la relación es sensible al tratamiento de los datos. Por lo tanto, aunque la asociación es robusta, el valor exacto de Pearson no debe considerarse completamente estable, pues depende de cómo se manejen los registros extremos y de las decisiones de limpieza del dataset.

¿Cambian las diferencias entre condiciones climáticas?
Si, cambian según cómo se mire. Sin controlar por distancia, la duración promedio varía muy poco entre climas (apenas 0.5 min de diferencia). Pero al controlar la distancia comparando solo viajes de 2 a 4 km, las diferencias se reducen todavía más y ya no muestran ningún patrón claro.

¿Por qué comparar todos los viajes directamente podría resultar engañoso?
Porque no todos los climas tienen la misma mezcla de distancias por decirlo asi. Si en un clima hay más viajes largos que en otro, ese clima va a mostrar duración promedio más alta, pero no por el clima en si, sino por la distancia. Comparar en bruto relaciones distintas y hacer parecer que el clima influye en esto no es muy adecuado, ya que en realidad es la distancia la que está jalando el resultado.


¿Qué papel puede estar desempeñando distancia km?
El de variable de confusión. Es la que realmente explica la mayor parte de la variación en duración (r= 0.72–0.89), y como probablemente varía también según el clima, genera una relación aparente entre clima y duración que en realidad no es directa.


¿Podemos concluir que el clima causa cambios en la duración del viaje?
No,al controlar por distancia, la asociación entre clima y duración se debilita hasta casi desaparecer. Con estos datos solo se puede hablar de una asociación muy débil, no de una relación causal. Eso sí esto solo aplica a la duración, lo único es que queda pendiente ver si el clima afecta la cantidad de viajes que la gente hace.
