```python
import pandas as pd
import numpy as np
df = pd.read_csv('bicicletas_compartidas_bogota.csv')
```

**Punto 1**

```python
sub = df[['distancia_km', 'duracion_min']].dropna()

x = sub['distancia_km'].to_numpy()
y = sub['duracion_min'].to_numpy()

r = sub['distancia_km'].corr(sub['duracion_min'])
print(f"Pearson r = {r:.4f}")
```

**1.b** 

**Dirección:** positiva, a mayor distancia recorrida, tiende a haber mayor duración del viaje.

**Intensidad:** moderada-alta. Un r de 0.72 no es una correlación perfecta, pero sí es considerablemente fuerte para datos de comportamiento humano/urbano, donde hay mucha variabilidad.

**Significado práctico:** el R² ≈ 0.52 indica que aproximadamente el 52% de la variabilidad en la duración de los viajes se puede explicar linealmente por la distancia recorrida. El 48% restante depende de otros factores: tipo de bicicleta (mecánica vs. eléctrica), clima, congestión, localidad, comportamiento del usuario (paradas, descansos), etc.

**1.c**

No es correcto afirmarlo únicamente con base en la correlación, porque esta muestra asociación, no causalidad. Aunque existe una relación física razonable entre distancia y duración, podrían influir otras variables como el clima o el tipo de bicicleta. Por eso, lo adecuado es decir que existe una asociación positiva entre ambas variables, no que la distancia causa directamente la duración.
